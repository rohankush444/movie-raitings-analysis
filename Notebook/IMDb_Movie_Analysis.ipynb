```python
df = df.drop_duplicates()

print(df.shape)
```

```python
df = df.fillna(0)
```

```python
df.describe()
```

```python
corr = df.corr(numeric_only=True)

corr
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(corr,
            annot=True,
            cmap="coolwarm",
            linewidths=0.5)

plt.title("Correlation Heatmap")

plt.savefig("Correlation_Heatmap.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.histplot(df['IMDB_Rating'], bins=20, kde=True)

plt.title("IMDb Rating Distribution")
plt.xlabel("IMDb Rating")
plt.ylabel("Number of Movies")

plt.savefig("Rating_Distribution.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
genre = df['Genre'].str.split(',', expand=True)[0]

top_genres = genre.value_counts().head(10)

plt.figure(figsize=(10,6))

sns.barplot(x=top_genres.values,
            y=top_genres.index)

plt.title("Top 10 Movie Genres")
plt.xlabel("Number of Movies")
plt.ylabel("Genre")

plt.savefig("Top10_Genres.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
genre = df.copy()

genre['Main_Genre'] = genre['Genre'].str.split(',').str[0]

avg_rating = genre.groupby('Main_Genre')['IMDB_Rating'].mean().sort_values(ascending=False)

plt.figure(figsize=(10,7))

sns.barplot(x=avg_rating.values,
            y=avg_rating.index)

plt.title("Average IMDb Rating by Genre")

plt.savefig("Genre_Average_Rating.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

```python
gross = df.copy()

gross['Gross'] = gross['Gross'].replace(',', '', regex=True)
gross['Gross'] = pd.to_numeric(gross['Gross'], errors='coerce')

top_gross = gross.sort_values(by='Gross', ascending=False).head(10)

plt.figure(figsize=(10,7))

sns.barplot(data=top_gross,
            x='Gross',
            y='Series_Title')

plt.title("Top 10 Highest Grossing Movies")

plt.savefig("Top_Grossing_Movies.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```
